# Part 3. Perform Data Partitioning

The dataset is divided into a training set (80%) and a testing set (20%) using `random_state=42` so that the split can be reproduced.

Two precautions are taken so that the testing results are not inflated:

1. **Duplicate records are removed before the split.** The data understanding step found 1,202 exact duplicate rows. In addition, `name` and `torque` are not used as model inputs, so a small number of records that differ only in those two columns become identical in every modeling feature and in the selling price. If any of these copies stayed in the data, the same record could land in both the training and the testing set, and the model would be tested on records it had already seen. Both kinds of duplicates are therefore removed first.
2. **Only record IDs are partitioned here.** The cleaning in Part 4 uses row-by-row operations only (removing duplicates, trimming text, and converting text specifications to numbers), so it can be applied to the whole dataset without using any information from the testing records. The same IDs are used in Part 5 to build `X_train`, `X_test`, `y_train`, and `y_test`. Anything that learns from the data (imputation, scaling, encoding) is fitted on the training records only, inside the pipeline.

```python
from sklearn.model_selection import train_test_split

# Step 1: remove exact duplicate rows
df_unique = df.drop_duplicates().reset_index(drop=True)
exact_removed = len(df) - len(df_unique)

# Step 2: remove records that are identical in every modeling feature and in the
# selling price (they differ only in name or torque, which are not model inputs)
modeling_columns = [c for c in df_unique.columns if c not in ("name", "torque")]
before = len(df_unique)
df_unique = df_unique.drop_duplicates(subset=modeling_columns).reset_index(drop=True)
same_features_removed = before - len(df_unique)

# Partition the record IDs into 80% training and 20% testing
train_ids, test_ids = train_test_split(
    df_unique.index,
    test_size=0.20,
    random_state=42
)

print("Original records:", len(df))
print("Exact duplicates removed:", exact_removed)
print("Records identical in all modeling features removed:", same_features_removed)
print("Records remaining:", len(df_unique))
print("Training records:", len(train_ids))
print("Testing records:", len(test_ids))
```

# Part 4. Data Cleaning

Data cleaning was performed to improve the quality and consistency of the dataset before applying the regression models. During the data understanding stage, duplicate records, missing values, and technical specifications stored as text were identified.

The cleaning process includes removing exact duplicate records, standardizing text values, converting selected vehicle specifications into numerical data, and checking missing values. The `torque` attribute will not be included in the initial regression model because its values use different formats and units. The `name` attribute will also be excluded from the initial model because it contains detailed vehicle names with many unique values.

Records that become identical after `name` and `torque` are excluded are removed, as in Part 3. Missing values are not filled at this stage. They are imputed inside the modeling pipeline in Part 5, using values calculated from the training records only, so that no information from the testing records is used during training. These steps help prepare the dataset for model training and evaluation while preserving useful information about the vehicles.

```python
# Create a copy so the original dataset remains unchanged
df_clean = df.copy()

# Check the dataset before cleaning
print("Original dataset shape:", df_clean.shape)
print("Exact duplicate rows:", df_clean.duplicated().sum())
print("Total missing values:", df_clean.isnull().sum().sum())

# Remove extra spaces from column names and text values
df_clean.columns = df_clean.columns.str.strip()

for col in [c for c in df_clean.columns if not pd.api.types.is_numeric_dtype(df_clean[c])]:
    df_clean[col] = df_clean[col].str.strip()

# Remove exact duplicate records
df_clean = df_clean.drop_duplicates().reset_index(drop=True)

print("\nDataset shape after removing duplicates:", df_clean.shape)
print("Remaining exact duplicates:", df_clean.duplicated().sum())
```

```python
# Convert vehicle specifications from text to numeric values
numeric_spec_columns = ["mileage", "engine", "max_power"]

for col in numeric_spec_columns:
    df_clean[col] = pd.to_numeric(
        df_clean[col].astype("string").str.extract(
            r"(\d+(?:\.\d+)?)"
        )[0],
        errors="coerce"
    ).astype("float64")

# Ensure seats is numeric
df_clean["seats"] = pd.to_numeric(
    df_clean["seats"], errors="coerce"
)

# Display the converted columns
print(df_clean[
    ["mileage", "engine", "max_power", "seats"]
].dtypes)

display(df_clean[
    ["mileage", "engine", "max_power", "seats"]
].head())
```

```python
# Check missing values after converting the text specifications
missing_after = df_clean.isnull().sum().to_frame("Missing Values")
display(missing_after)

# The missing values are not filled here. Medians and modes calculated on the
# whole dataset would carry information from the testing records into training,
# so imputation is done inside the Part 5 pipeline and fitted on the training
# records only.
print("\nDataset shape after cleaning:", df_clean.shape)
```

```python
# Create the dataset for the regression models
df_model = df_clean.drop(columns=["name", "torque"])

# Remove records that became identical in every modeling feature and in the price
# (the same step that was applied before the partition in Part 3)
before_model_dedup = len(df_model)
df_model = df_model.drop_duplicates().reset_index(drop=True)
print("Records identical in all modeling features removed:", before_model_dedup - len(df_model))

# The model records must follow the same order as the records partitioned in Part 3
assert len(df_model) == len(df_unique)
assert (df_model[["year", "selling_price", "km_driven"]].values
        == df_unique[["year", "selling_price", "km_driven"]].values).all()
print("Row order matches the Part 3 partition.")

# Confirm that the target variable is available
print("\nTarget variable: selling_price")
print("Model dataset shape:", df_model.shape)

# Check the final data types
display(df_model.dtypes.to_frame("Data Type"))

# Count the missing values that remain
print("Missing values to be imputed in the Part 5 pipeline:", df_model.isnull().sum().sum())

# Preview the cleaned modeling dataset
display(df_model.head())
```

# Part 5. Establishing the Machine Learning Pipeline (Preprocessing and Modelling)

The cleaned dataset is divided into 80% training data and 20% testing data using the record IDs created in Part 3. Preprocessing is defined once as a `ColumnTransformer` and placed at the start of every model pipeline, so it is fitted on the training records only:

- **Numerical features** (`year`, `km_driven`, `mileage`, `engine`, `max_power`, `seats`): missing values are filled with the median, then the values are standardized.
- **Categorical features** (`fuel`, `seller_type`, `transmission`, `owner`): missing values are filled with the most frequent value, then the categories are one-hot encoded.

Five regression models are trained to predict used-car selling prices: LASSO, Ridge, Elastic Net, Regression Tree, and Random Forest.

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import Lasso, Ridge, ElasticNet
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Build the feature matrix and target from the cleaned dataset
X = df_model.drop(columns=["selling_price"])
y = df_model["selling_price"]

# Apply the partition created in Part 3
X_train, y_train = X.loc[train_ids], y.loc[train_ids]
X_test, y_test = X.loc[test_ids], y.loc[test_ids]

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)

numeric_features = ["year", "km_driven", "mileage", "engine", "max_power", "seats"]
categorical_features = ["fuel", "seller_type", "transmission", "owner"]

preprocessor = ColumnTransformer([
    ("num", Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]), numeric_features),
    ("cat", Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore"))
    ]), categorical_features)
])
```

## Part 5.1. Machine Learning LASSO Model Training and Testing

```python
# Train the LASSO model
lasso_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", Lasso(alpha=1.0, max_iter=10000))
])

lasso_model.fit(X_train, y_train)

# Test the model
lasso_predictions = lasso_model.predict(X_test)

# Evaluate model performance
lasso_mae = mean_absolute_error(y_test, lasso_predictions)
lasso_rmse = np.sqrt(mean_squared_error(y_test, lasso_predictions))
lasso_r2 = r2_score(y_test, lasso_predictions)

print("LASSO Model Performance")
print(f"MAE:  {lasso_mae:.3f}")
print(f"RMSE: {lasso_rmse:.3f}")
print(f"R²:   {lasso_r2:.3f}")
```

**Summary:** LASSO was trained with alpha = 1.0 on the training records and then tested on the 1,382 testing records. It produced an MAE of 170,697, an RMSE of 297,063, and an R² of 0.601, so it explains about 60% of the variation in selling prices on unseen vehicles. It has the lowest RMSE of the three linear models, but only by a small margin, and it is far behind the tree-based models. This fits the fact that a linear model cannot follow non-linear price patterns.

## Part 5.2. Machine Learning Ridge Model Training and Testing

```python
ridge_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", Ridge(alpha=1.0))
])

ridge_model.fit(X_train, y_train)
ridge_predictions = ridge_model.predict(X_test)

ridge_mae = mean_absolute_error(y_test, ridge_predictions)
ridge_rmse = np.sqrt(mean_squared_error(y_test, ridge_predictions))
ridge_r2 = r2_score(y_test, ridge_predictions)

print("Ridge Model Results")
print("MAE:", ridge_mae)
print("RMSE:", ridge_rmse)
print("R²:", ridge_r2)
```

**Summary:** Ridge (alpha = 1.0) produced an MAE of 170,996, an RMSE of 297,221, and an R² of 0.601. The results are almost identical to LASSO, so shrinking the coefficients instead of removing some of them made practically no difference here.

## Part 5.3. Machine Learning Elastic Net Model Training and Testing

```python
elastic_net_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", ElasticNet(alpha=1.0, max_iter=10000))
])

elastic_net_model.fit(X_train, y_train)
elastic_net_predictions = elastic_net_model.predict(X_test)

elastic_net_mae = mean_absolute_error(y_test, elastic_net_predictions)
elastic_net_rmse = np.sqrt(
    mean_squared_error(y_test, elastic_net_predictions)
)
elastic_net_r2 = r2_score(y_test, elastic_net_predictions)

print("Elastic Net Model Results")
print("MAE:", elastic_net_mae)
print("RMSE:", elastic_net_rmse)
print("R²:", elastic_net_r2)

```

**Summary:** Elastic Net (alpha = 1.0, default `l1_ratio` of 0.5) produced an MAE of 158,887, an RMSE of 305,081, and an R² of 0.579. Its MAE is the lowest of the linear models, but its RMSE and R² are the weakest of the three, which suggests that typical errors are somewhat smaller while the errors on expensive vehicles are larger.

## Part 5.4. Machine Learning Regression Tree Model Training and Testing

```python
tree_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", DecisionTreeRegressor(
        max_depth=10,
        random_state=42
    ))
])

tree_model.fit(X_train, y_train)
tree_predictions = tree_model.predict(X_test)

tree_mae = mean_absolute_error(y_test, tree_predictions)
tree_rmse = np.sqrt(mean_squared_error(y_test, tree_predictions))
tree_r2 = r2_score(y_test, tree_predictions)

print("Regression Tree Model Results")
print("MAE:", tree_mae)
print("RMSE:", tree_rmse)
print("R²:", tree_r2)
```

**Summary:** The regression tree (`max_depth` = 10) produced an MAE of 87,158, an RMSE of 158,597, and an R² of 0.886. This is a large improvement over the linear models, which indicates that splitting vehicles into groups captures price patterns that a straight-line model misses. A single tree can be unstable, so the ensemble in the next section is a useful comparison.

## Part 5.5. Machine Learning Random Forest Model Training and Testing

```python
random_forest_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", RandomForestRegressor(
        n_estimators=100,
        max_depth=15,
        random_state=42,
        n_jobs=-1
    ))
])

random_forest_model.fit(X_train, y_train)
random_forest_predictions = random_forest_model.predict(X_test)

random_forest_mae = mean_absolute_error(
    y_test, random_forest_predictions
)
random_forest_rmse = np.sqrt(
    mean_squared_error(y_test, random_forest_predictions)
)
random_forest_r2 = r2_score(
    y_test, random_forest_predictions
)

print("Random Forest Model Results")
print("MAE:", random_forest_mae)
print("RMSE:", random_forest_rmse)
print("R²:", random_forest_r2)
```

**Summary:** The random forest (100 trees, `max_depth` = 15) produced an MAE of 74,693, an RMSE of 133,543, and an R² of 0.919. It has the best result on all three metrics. Averaging many trees reduces the instability of a single tree, which is a likely reason it improves on the regression tree.

### Key Findings
- **Random Forest performed best.** It reached an R² of 0.919, meaning it explained about 91.9% of the variation in selling prices in the testing set. The regression tree came second with an R² of 0.886.
- **Tree-based models clearly outperformed the linear models.** LASSO, Ridge, and Elastic Net stayed between 0.579 and 0.601. The random forest's MAE of 74,693 is less than half of the lowest linear MAE (158,887).
- **The linear models behave similarly.** The choice of penalty (LASSO, Ridge, or Elastic Net) mattered much less than the choice between a linear and a tree-based model. This suggests that the price relationships, such as the effect of age, power, and engine size, are not well described by straight-line terms.
- **Converting the text specifications to numbers made them usable.** `mileage`, `engine`, and `max_power` were stored as text with units and were converted to numeric columns in Part 4, so that the models could use them.
- **Practical meaning.** The random forest's average error of 74,693 is roughly 19% of the median selling price in the cleaned dataset (400,000). It can serve as a reference price for comparing listings, not as an exact valuation.

### Limitations
- The scores come from a single 80/20 split with fixed hyperparameters. Cross-validation and hyperparameter tuning were not performed, so the results may change somewhat with a different split.
- Different listings of the same vehicle model can still have similar features and prices, so some similarity between the training and testing records remains.
- A few vehicles have a `mileage` or `max_power` of 0 in the source data. These were kept as recorded.
- `mileage` combines values in km/l and km/kg, and only the number was kept.
- `name` and `torque` were excluded, and vehicle condition, accident history, and location are not in the dataset.
- The results show associations in historical listings. They do not show that any attribute causes a price change.
